```python
#now lets assing customer_id to credit table

credit["customer_id"]=np.random.choice(
    customer_ids,
    size=len(credit),
    replace=True
)
```

```python
#lets assing the customer_id to transactions table
transactions["customer_id"] = np.random.choice(
    customer_ids,
    size=len(transactions),
    replace=True
)
```

```python
#lets aggregate credit table to customer level

credit_agg=credit.groupby("customer_id").agg(
    total_credit_records=("SeriousDlqin2yrs","count"),
    avg_debt_ratio=("DebtRatio","mean"),
    max_90_days_late=("NumberOfTimes90DaysLate","max"),
    total_late_events=("NumberOfTime30-59DaysPastDueNotWorse","sum"),
    ever_defaulted=("SeriousDlqin2yrs","max")).reset_index()
```

```python
credit_agg.head()

```

```python
credit_agg.shape
```

# Lets make dim_customer table

```python
dim_customer = customers.merge(
    credit_agg,
    on="customer_id",
    how="left")
dim_customer.head()


               
```

```python
dim_customer.shape
```

```python
credit_cols=[
     "total_credit_records",
   "avg_debt_ratio",
    "max_90_days_late",
    "total_late_events",
    "ever_defaulted"
]
dim_customer[credit_cols]=dim_customer[credit_cols].fillna(0)
    
```

```python
dim_customer.head()
```

```python
dim_customer.to_parquet("../data/processed/dim_customer.parquet")
pd.read_parquet("../data/processed/dim_customer.parquet").shape
```

```python
fact_loans = loans[
[
    "customer_id",
    "SK_ID_CURR",
    "AMT_CREDIT",
    "AMT_ANNUITY",
    "AMT_GOODS_PRICE",
    "AMT_INCOME_TOTAL",
    "TARGET"
]
]
    
```

```python
fact_transactions =transactions.groupby("customer_id").agg(
        txn_count=("TransactionID", "count"),
        total_spend=("TransactionAmount (INR)", "sum"),
        avg_txn_amount=("TransactionAmount (INR)", "mean"),
        max_txn_amount=("TransactionAmount (INR)", "max"),
        last_account_balance=("CustAccountBalance", "last")).reset_index()
```

```python
fact_transactions.isna().sum()
```

```python
fact_transactions.to_parquet(
    "../data/processed/fact_transactions.parquet",
    index=False
)
pd.read_parquet("../data/processed/fact_transactions.parquet").shape
```
